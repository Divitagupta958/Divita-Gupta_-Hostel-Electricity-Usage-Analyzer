## 2. Load your dataset

```python
df = pd.read_csv("Student_16_dataset.csv")
display(df.head())
```

## 3. Explore the data
Use `shape`, `columns`, `info()`, `isnull().sum()` and `duplicated().sum()` to inspect the dataset.


```python
print("Datasheet shape:", df.shape)
print("\ncolumns:", df.columns.tolist())
print("\nDatasheet Info:")
df.info()
print("\nMissing values per column:\n", df.isnull().sum())
print("\nTotal duplicate rows:", df.duplicated().sum())
```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
import numpy as np
#select numeric columns and convert to a numpy array
numeric_df = df.select_dtypes(include=[np.number])
numeric_array = numeric_df.to_numpy()
print("Numeric columns selected:", numeric_df.columns.tolist())
print("Numpy Array:\n", numeric_array[:5]) 
print("\nMean values:", np.mean(numeric_array, axis=0))
print("\nMax values:", np.max(numeric_array, axis=0))
print("\nMin values:", np.min(numeric_array, axis=0))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
#1. Create calculated columns
df['Units_Per_Occupant'] = df['Units_Used'] / df['Occupants']
df['Daily_Units'] = df['Units_Used'] / 30
df['Estimated_Bill'] = df['Units_Used'] * 8

#2. Filter records using a condition
high_usage_rooms = df[df['Units_Used'] > 300]
if len(high_usage_rooms) > 0:
    print("\nHigh Usage Rooms Found:", len(high_usage_rooms))
    print(high_usage_rooms)
else:
    print("\nNo rooms exceeded 300 units.")
# 3. Sort Records and display the top/bottom results
sorted_df = df.sort_values(by='Units_Used', ascending=False)
print("\nTop 5 Highest Usage Records:")
print(sorted_df.head())
print("\nBottom 5 Lowest Usage Records:")
print(sorted_df.tail())
```

## 6. Your individual project challenge
**Challenge:** Calculate units per occupant; identify highest-use room and compare blocks.

Write your solution below and explain what the result means.


```python
#1. Calculate units per occupant
df['Units_Per_Occupant'] = df['Units_Used'] / df['Occupants']
#2. Identify the Highest-Use Room
highest_use_room = df.sort_values(by='Units_Used', ascending=False).head(1)
print("Highest_Use Room Details:")
print(highest_use_room)
print("-" * 40)
#3. Compare blocks(total and average electricity usage)
block_comparison = df.groupby('Block').agg(
    Total_Rooms=('Room_ID', 'count'),
    Total_Units=('Units_Used', 'sum'),
    Avg_Units_Per_Room=('Units_Used', 'mean'),
    Avg_Units_Per_Occupant=('Units_Per_Occupant', 'mean')
)
print("\n---Block Comparison---")
print(block_comparison)



```
